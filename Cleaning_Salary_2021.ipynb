```python
#minmax normalization (way1)
max_Salary=df['Salary'].max()
min_Salary=df['Salary'].min()
min_max_normalization=(df['Salary']-min_Salary)/(max_Salary-min_Salary)
min_max_normalization.head()
```

```python
#minmax normalization (way2)
from sklearn.preprocessing import MinMaxScaler
df['min_max_normalization']=MinMaxScaler().fit_transform(df[['Salary']])
df['min_max_normalization'].head()
```

```python
#z_score (way1)
Salary_mean=df['Salary'].mean()
Salary_Std=df['Salary'].std()
Z_Score_Normalization=(df['Salary']-Salary_mean)/Salary_Std
Z_Score_Normalization.head()
```

```python
#z_Score (way2)
from sklearn.preprocessing import StandardScaler
df['Z_Score_normalization']=StandardScaler().fit_transform(df[['Salary']])
df['Z_Score_normalization'].head()
```

```python
#robust normalization (way1)
Q1=df['Salary'].quantile(0.25)
Q3=df['Salary'].quantile(0.75)
IQR=Q3-Q1
Salary_medin=df['Salary'].median()
robust_normalization=(df['Salary']-Salary_medin)/IQR
robust_normalization.head()
```

```python
#Robust Normalization (way2)
from sklearn.preprocessing import RobustScaler
df['Robust Normalization']=RobustScaler().fit_transform(df[['Salary']])
df['Robust Normalization'].head()
```

```python
#decimal Normalization
Max_Value=abs(df['Salary'].max())
length=len(str(Max_Value))
df['decimal_Normalization']=df['Salary']/(10**length)
df['decimal_Normalization'].head()

```
